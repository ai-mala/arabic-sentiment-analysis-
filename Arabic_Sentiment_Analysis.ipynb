```python
# الجمل وتصنيف مشاعرها
texts = [
    ("الخدمة رائعة جداً", "إيجابي"),
    ("التجربة سيئة ومزعجة", "سلبي"),
    ("المنتج عادي", "محايد")
]

# عرض الجمل مع تصنيفها
for text, sentiment in texts:
    print("الجملة:", text)
    print("المشاعر:", sentiment)
    print("-" * 30)
```

```python
# كلمات تدل على المشاعر
positive_words = ["رائع", "ممتاز", "جميل", "ممتعة", "رائعة"]
negative_words = ["سيئ", "سيئة", "مزعج", "مزعجة", "سيئة"]

# دالة لتحليل المشاعر
def analyze_sentiment(text):
    if any(word in text for word in positive_words):
        return "إيجابي 😊"
    elif any(word in text for word in negative_words):
        return "سلبي 😞"
    else:
        return "محايد 😐"

# تجربة تحليل الجمل
sentences = [
    "الخدمة رائعة جداً",
    "التجربة سيئة ومزعجة",
    "المنتج عادي"
]

for sentence in sentences:
    result = analyze_sentiment(sentence)
    print("الجملة:", sentence)
    print("النتيجة:", result)
    print("-" * 30)
```

```python
# استيراد الأدوات اللازمة
from transformers import pipeline

# تحميل نموذج تحليل المشاعر العربية
model = pipeline(
    "sentiment-analysis",
    model="CAMeL-Lab/bert-base-arabic-camelbert-da-sentiment"
)

print("تم تحميل نموذج تحليل المشاعر بنجاح!")
```

```python
# جمل جديدة لاختبار النموذج
sentences = [
    "الخدمة رائعة وأنا سعيدة جداً",
    "التجربة سيئة ولن أكررها",
    "المنتج ممتاز وأنصح الجميع به",
    "أنا غير راضية عن الخدمة"
]

# تحليل المشاعر باستخدام نموذج الذكاء الاصطناعي
results = model(sentences)

# عرض النتائج
for sentence, result in zip(sentences, results):
    print("الجملة:", sentence)
    print("التصنيف:", result["label"])
    print("نسبة الثقة:", round(result["score"] * 100, 2), "%")
    print("-" * 40)
```

```python
import pandas as pd

# تحويل نتائج النموذج إلى جدول
df = pd.DataFrame({
    "الجملة": sentences,
    "التصنيف": [result["label"] for result in results],
    "نسبة الثقة": [
        round(result["score"] * 100, 2)
        for result in results
    ]
})

# حفظ النتائج في ملف CSV
df.to_csv("sentiment_results.csv", index=False, encoding="utf-8-sig")

# عرض الجدول
df
```
